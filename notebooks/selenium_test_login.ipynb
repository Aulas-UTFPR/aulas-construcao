# Importar Dependências

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import unittest
import time
```

# Caso de Teste

```python
class TestLogin(unittest.TestCase):
    def setUp(self):
        # Setup do navegador (Chrome)
        self.driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
        self.driver.get("https://the-internet.herokuapp.com/login")
        self.driver.maximize_window()
        time.sleep(1)  # Pequena pausa para carregamento

    def test_login_success(self):
        driver = self.driver
        driver.find_element(By.ID, "username").send_keys("tomsmith")
        driver.find_element(By.ID, "password").send_keys("SuperSecretPassword!")
        driver.find_element(By.CLASS_NAME, "radius").click()
        time.sleep(1)
        success_message = driver.find_element(By.ID, "flash").text
        self.assertIn("You logged into a secure area!", success_message)

    def test_login_failure(self):
        driver = self.driver
        driver.find_element(By.ID, "username").send_keys("wronguser")
        driver.find_element(By.ID, "password").send_keys("wrongpass")
        driver.find_element(By.CLASS_NAME, "radius").click()
        time.sleep(1)
        error_message = driver.find_element(By.ID, "flash").text
        self.assertIn("Your username is invalid!", error_message)

    def tearDown(self):
        self.driver.quit()
```

```python
unittest.main(argv=[''], verbosity=2, exit=False)
```
